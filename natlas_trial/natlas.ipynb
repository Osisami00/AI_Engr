```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
from datetime import datetime
import os
from dotenv import load_dotenv

load_dotenv()

model_name = "NCAIR1/N-ATLaS"
api = os.getenv("read_access_key")  # your Hugging Face token

# Pass token for gated model
tokenizer = AutoTokenizer.from_pretrained(model_name, use_auth_token=api)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    use_auth_token=api,
    torch_dtype=torch.float16,
    device_map="auto"
)

def format_text_for_inference(messages):
    current_date = datetime.now().strftime('%d %b %Y')
    text = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=False,
        date_string=current_date
    )
    return text

# Example usage
q_chat = [
    {'role':'system','content':'you are a large language model trained by Awarri AI technologies. You are a friendly assistant and you are here to help.'},
    {'role': 'user', 'content': "menene ake nufi da gwagwarmaya"}
]

text = format_text_for_inference(q_chat)

# Tokenize input text
input_tokens = tokenizer(text, return_tensors='pt').to('CUDA')

# Generate response
outputs = model.generate(
    input_ids=input_tokens['input_ids'],
    max_new_tokens=1000,
    use_cache=True,
    repetition_penalty=1.12,
    temperature=0.1
)

# Decode output
text = tokenizer.batch_decode(outputs, skip_special_tokens=True)[0]
print(text)
```
