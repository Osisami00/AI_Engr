# 1. Data Preparation

a. Turn data into numbers - numerical encoding


b. Build model to learn patterns in number

```python
# Create Known parameters
weight = 0.7
bias = 0.3

# creatte data 
start = 0
end = 1
step = 0.02

X = torch .arange(start, end, step).unsqueeze(dim=1)
y = weight * X + bias

X[:10], y[:10]
```

X == data

y = weight * X + bias

# Building a model that can learn relationship

X(features) and y (labels)

a. Data Splitting

```python
# Create train/test split

train_split = int(0.8 * len(X))  #80percent of the data is used for training
X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]

len(X_train), len(y_train), len(X_test), len(y_test)
```

Let's visualize training and testing dataset

```python
def plot_prediction(train_data=X_train,
                    train_labels=y_train,
                    test_data=X_test,
                    test_labels=y_test,
                    predictions = None                    
                    ):
    
    """
    Plots training data, test data and compares predictions.
    """


    plt.figure(figsize=(10,7))

    # plot training data in blue
    plt.scatter(train_data, train_labels, c="b", s=20, label="Training Data")


    # Plot test data in green
    plt.scatter(test_data, test_labels, c="g", s=20, label="Testing Data")

    if predictions is not None:
        # Plot predictions in red (predictions were made on the test data)
        plt.scatter(test_data, predictions, c="r", s=20, label="Predictions")


    # show legend
    plt.legend(prop={"size":14});
```

```python
plot_prediction();
```
