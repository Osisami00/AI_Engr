Tables

````python
text = "Tesla is an electric vehicle company. Its stock price has increased by 20% in the last year.Apple is a technology company known for its iphones. Its stock price has increased by 15% in the last year. Amazon is an e-commerce and cloud computing giant. Its stock price has increased by 10% in thelast year."

persona = f"""Convert the following information into a table format (in staright lines not broken lines) with columns for company , industry
and stock performance:
Text: {text}' ** """

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents= persona
)


print(f"{response.text}")
````
